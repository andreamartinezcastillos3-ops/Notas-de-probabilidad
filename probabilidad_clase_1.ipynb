# EXPERIMENTOS

##Experimento determinista

Un experimento determinista es aquel que produce siempre el mismo resultado cuando se repite bajo las mismas condiciones iniciales. Es decir, si las variables que intervienen en el experimento se mantienen constantes, el resultado será siempre el mismo.



1.   Ejemplo de area



```python
#Logica del area
import sympy

def area(radio):
    return sympy.pi * (radio ** 2)
```

```python
area(10)
```

```python
import sympy

def area_numerico(radio):
    return 3.1416 * (radio ** 2)
```

## Experimento Aleatorio

Un experimento aleatorio es aquel que, aún cuando se repita bajo las mismas condiciones, no produce siempre el mismo resultado y no es posible predecir con certeza cuál ocurrirá.

2. Ejemplo de moneda

```python
#Logica de lanzamiento de una moneda
import random

moneda_lados = ["aguila", "sol"]

def lanzar_moneda():
    return random.choice(moneda_lados)

lanzar_moneda()
```

```python
#Logica de lanzamiento de un dado
import random

dado_caras = ["1", "2", "3", "4", "5", "6"]

def lanzar_dado():
    return random.choice(dado_caras)

lanzar_dado()
```
